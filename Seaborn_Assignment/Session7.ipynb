TASK:2.
Using Seaborn, plot a heatmap of the correlation matrix you generated for the food delivery data, and add annotations to show the correlation values inside each cell.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
data = {
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(100, 1000, 20),
    'tip_amount': np.random.randint(10, 100, 20),
    'rating': np.random.randint(1, 6, 20),
    'delivery_distance': np.random.uniform(1, 15, 20).round(2)
}
```

```python
df = pd.DataFrame(data)
```

```python
correlation_matrix = df.corr()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Food Delivery Orders - Correlation Heatmap')
plt.tight_layout()
plt.show()
```

TASK:3.
Change the color palette of your Seaborn heatmap to a diverging palette (such as 'coolwarm' or 'RdYlGn') to better visualize positive and negative correlations, and explain which two features in your data have the strongest negative correlation.<br><br><em><strong>Hint:</strong> Use the cmap parameter in sns.heatmap().</em>

```python
np.random.seed(42)
```

```python

data = {
    'delivery_time': np.random.randint(15, 60, 20),
    'order_value': np.random.randint(100, 1000, 20),
    'tip_amount': np.random.randint(10, 100, 20),
    'rating': np.random.randint(1, 6, 20),
    'delivery_distance': np.random.uniform(1, 15, 20).round(2)
}
```

```python
df = pd.DataFrame(data)
```

```python
correlation_matrix = df.corr()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='RdYlGn',
    fmt='.2f',
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title('Food Delivery Correlation Heatmap')
plt.tight_layout()
plt.show()

```

```python
# Find the strongest negative correlation
corr_pairs = correlation_matrix.where(
    np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool)
).stack()

strongest_negative = corr_pairs.idxmin()
negative_value = corr_pairs.min()

print("Strongest negative correlation:")
print(strongest_negative)
print("Correlation value:", round(negative_value, 2))
```
