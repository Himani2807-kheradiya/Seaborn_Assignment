TASK:1.
Load a CSV file containing Flipkart product prices and use sns.boxplot() to visualize the price distribution.


```python

import pandas as pd

data = {
    'Product': [
        'Mouse', 'Keyboard', 'Headphones', 'Smartwatch',
        'Speaker', 'Monitor', 'Mobile', 'Tablet',
        'Laptop', 'Charger', 'Power Bank', 'Camera',
        'Earbuds', 'Printer', 'Gaming Console'
    ],
    'Price': [
        499, 899, 1499, 2499, 1999,
        8999, 14999, 18999, 54999, 399,
        1299, 25999, 1799, 7999, 39999
    ]
}

df = pd.DataFrame(data)
df.to_csv('flipkart_products.csv', index=False)

print("CSV file created successfully!")
print(df.head())
```

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('flipkart_products.csv')
```

```python
print(df.head())
```

```python
sns.boxplot(data=df, y='Price')

plt.title('Distribution of Flipkart Product Prices')
plt.ylabel('Price (₹)')

plt.show()
```

TASK:2.
Use sns.violinplot() to compare the order amounts for two different Zomato restaurant categories (e.g., 'Fast Food' vs 'Desserts') from a sample dataset.

```python
data = {
    'Category': [
        'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food',
        'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food',
        'Desserts', 'Desserts', 'Desserts', 'Desserts', 'Desserts',
        'Desserts', 'Desserts', 'Desserts', 'Desserts', 'Desserts'
    ],
    'Order_Amount': [
        150, 200, 250, 180, 300, 220, 350, 170, 280, 400,
        100, 120, 150, 180, 200, 130, 160, 250, 220, 140
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df.head())
```

```python
sns.violinplot(data=df, x='Category', y='Order_Amount')

plt.title('Zomato Order Amounts by Restaurant Category')
plt.xlabel('Restaurant Category')
plt.ylabel('Order Amount (₹)')

plt.show()
```
