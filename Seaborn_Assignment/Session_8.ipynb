TASK:1.
Use sns.scatterplot() to visualize the relationship between the number of followers and the number of posts for 20 fictional Instagram users. Label the axes appropriately.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import random
```

```python
data = {
    'User': [f'User{i}' for i in range(1, 21)],
    'Followers': [
        120, 450, 800, 1500, 2300,
        3200, 4100, 5200, 6500, 7800,
        9000, 11000, 12500, 14000, 16000,
        18500, 21000, 25000, 28000, 32000
    ],
    'Posts': [
        25, 60, 90, 120, 150,
        180, 210, 240, 270, 300,
        330, 360, 390, 420, 450,
        480, 510, 550, 590, 630
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x='Followers',
    y='Posts',
    color='purple',
    s=80
)

plt.title('Instagram Followers vs Number of Posts')
plt.xlabel('Number of Followers')
plt.ylabel('Number of Posts')

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

TASK:2.
Create a line plot using sns.lineplot() to show the daily number of food orders on Zomato for each day over a month. Use a list of 30 random order counts to simulate the data.

```python
random.seed(42)
orders = [random.randint(100, 500) for _ in range(30)]
```

```python
df = pd.DataFrame({
    'Day': list(range(1, 31)),
    'Orders': orders
})
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x='Day',
    y='Orders',
    marker='o',
    color='red'
)

plt.title('Daily Zomato Food Orders Over a Month')
plt.xlabel('Day of the Month')
plt.ylabel('Number of Orders')
plt.xticks(range(1, 31, 2))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

TASK:3.
Using sns.relplot(), plot the relationship between movie ratings and the number of reviews for different genres (Action, Comedy, Drama) based on a sample dataset of 15 movies. Use the 'hue' parameter to differentiate genres.<br><br><em><strong>Hint:</strong> Create a DataFrame with columns: 'movie', 'rating', 'reviews', and 'genre'.</em>

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'movie': [
        'Jawan', 'Pathaan', 'War', 'Dangal', 'Singham',
        '3 Idiots', 'Munna Bhai M.B.B.S.', 'Hera Pheri',
        'Chup Chup Ke', 'Welcome',
        '12th Fail', 'Dangal', 'Taare Zameen Par',
        'Lagaan', 'Bajrangi Bhaijaan'
    ],
    'rating': [
        7.0, 5.8, 6.5, 8.3, 6.8,
        8.4, 8.1, 8.2, 6.9, 7.4,
        8.7, 8.3, 8.3, 8.1, 8.1
    ],
    'reviews': [
        2200, 1800, 1600, 2500, 1200,
        2800, 2000, 1900, 1100, 1700,
        2300, 2500, 2100, 1800, 2400
    ],
    'genre': [
        'Action', 'Action', 'Action', 'Action', 'Action',
        'Comedy', 'Comedy', 'Comedy', 'Comedy', 'Comedy',
        'Drama', 'Drama', 'Drama', 'Drama', 'Drama'
    ]
}

df = pd.DataFrame(data)

print(df)

sns.relplot(
    data=df,
    x='rating',
    y='reviews',
    hue='genre',
    kind='scatter',
    s=100
)

plt.title('Hindi Movies: Ratings vs Number of Reviews')
plt.xlabel('Movie Rating')
plt.ylabel('Number of Reviews')
plt.show()
```

TASK:4.
Generate a regression plot using sns.regplot() to analyze the correlation between the number of hours spent on YouTube and exam scores for a group of 25 students. Interpret whether the trend is positive or negative.

```python
data = {
    'YouTube_Hours': [
        1, 2, 3, 1, 4, 2, 5, 3, 6, 2,
        4, 7, 1, 5, 3, 8, 2, 6, 4, 7,
        3, 9, 5, 8, 6
    ],
    'Exam_Scores': [
        95, 90, 85, 92, 78, 88, 72, 83, 65, 89,
        76, 60, 96, 70, 84, 55, 91, 64, 79, 58,
        82, 48, 73, 52, 66
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x='YouTube_Hours',
    y='Exam_Scores',
    color='blue',
    scatter_kws={'s': 60},
    line_kws={'color': 'red'}
)

plt.title('YouTube Hours vs Exam Scores')
plt.xlabel('Hours Spent on YouTube per Day')
plt.ylabel('Exam Scores (%)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()


```

```python

correlation = df['YouTube_Hours'].corr(df['Exam_Scores'])
print("Correlation:", round(correlation, 2))
```
