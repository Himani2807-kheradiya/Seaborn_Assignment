TASK:1.
Create a Seaborn barplot showing the number of orders per food category (e.g., Pizza, Burger, Biryani, Momos, Ice Cream) and save the chart as a PNG file with 300 DPI.

```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    'Food Category': ['Pizza', 'Burger', 'Biryani', 'Momos', 'Ice Cream'],
    'Number of Orders': [150, 120, 180, 100, 90]
}

df = pd.DataFrame(data)
```

```python
sns.set_theme(style='whitegrid', palette='deep')
```

```python
sns.set_theme(style='whitegrid', palette='deep')

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=df,
    x='Food Category',
    y='Number of Orders'
)

plt.title('Number of Orders per Food Category', fontsize=16)
plt.xlabel('Food Category', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.tight_layout()

plt.savefig('food_category_orders.png', dpi=300, bbox_inches='tight')

plt.show()
```

TASK:2.
Generate a Seaborn line plot visualizing daily step counts for one week, then export the chart as an SVG file suitable for use in a PowerPoint presentation.<br><br><em><strong>Hint:</strong> Use plt.savefig('filename.svg', format='svg').</em>


```python
data = {
    'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
            'Friday', 'Saturday', 'Sunday'],
    'Steps': [4500, 6000, 7500, 5000, 8500, 10000, 7000]
}

df = pd.DataFrame(data)
```

```python
sns.set_theme(style='whitegrid', palette='deep')
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x='Day',
    y='Steps',
    marker='o',
    linewidth=2.5
)

plt.title('Daily Step Counts for One Week', fontsize=16)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Number of Steps', fontsize=12)
plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig('weekly_steps.svg', format='svg', bbox_inches='tight')

plt.show()
```

TASK:3.
Customize a Seaborn chart using the 'darkgrid' style and a color palette of your choice, then save it as a high-resolution PNG image with 1200 DPI.


```python
data = {
    'Product': ['Mobile', 'Laptop', 'Headphones', 'Tablet', 'Smartwatch'],
    'Sales': [85000, 120000, 45000, 65000, 35000]
}

df = pd.DataFrame(data)
```

```python
sns.set_theme(style='darkgrid', palette='bright')

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=df,
    x='Product',
    y='Sales'
)

plt.title('Product Sales Dashboard', fontsize=16, fontweight='bold')
plt.xlabel('Product', fontsize=12)
plt.ylabel('Sales Amount (₹)', fontsize=12)

for container in ax.containers:
    ax.bar_label(container, padding=5, fmt='%.0f')

plt.tight_layout()

plt.savefig(
    'product_sales_1200dpi.png',
    format='png',
    dpi=1200,
    bbox_inches='tight'
)

plt.show()
```

TASK:4.
Create a template function called apply_presentation_theme() that applies a consistent Seaborn style, font, and color palette to any chart, then use it to style and export a pie chart showing your top 5 most-used apps.

```python
def apply_presentation_theme():
    sns.set_theme(
        style='whitegrid',
        context='talk',
        palette='deep',
        font='DejaVu Sans',
        font_scale=0.8
    )

# Apply the presentation theme
apply_presentation_theme()
```

```python
apps = ['Instagram', 'YouTube', 'WhatsApp', 'Spotify', 'Zomato']
usage = [12, 10, 7, 5, 3]

plt.figure(figsize=(9, 7))

plt.pie(
    usage,
    labels=apps,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette('deep', 5),
    explode=(0.05, 0.05, 0, 0, 0),
    shadow=False,
    textprops={'fontsize': 11}
)

plt.title('Top 5 Most-Used Apps', fontsize=18, fontweight='bold')
plt.tight_layout()

plt.savefig(
    'top_5_apps_presentation.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

TASK:5.
Use ChatGPT to generate a code snippet for exporting a Seaborn heatmap as a 1200 DPI SVG file, then test and adapt the code to save your own chart.<br><br><em><strong>Hint:</strong> Paste your prompt and the code you received as a comment in your script.</em>

```python
data = {
    'Instagram': [5, 4, 3, 6, 5],
    'YouTube': [4, 5, 6, 4, 3],
    'WhatsApp': [3, 4, 5, 3, 4],
    'Spotify': [2, 3, 4, 5, 4],
    'Zomato': [1, 2, 3, 2, 1]
}

df = pd.DataFrame(
    data,
    index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
)
```

```python
sns.set_theme(style='white', palette='deep')
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    df,
    annot=True,
    cmap='YlGnBu',
    linewidths=0.5,
    fmt='d'
)

plt.title('Weekly App Usage Heatmap', fontsize=16)
plt.xlabel('Apps')
plt.ylabel('Day of the Week')

plt.tight_layout()

plt.savefig(
    'app_usage_heatmap.svg',
    format='svg',
    dpi=1200,
    bbox_inches='tight'
)

plt.show()
```
