TASK:1.
Import Seaborn and use sns.histplot() to visualize the distribution of delivery times (in minutes) from a sample Swiggy order dataset with at least 30 values.

```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
delivery_times = [
    32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
    27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
    36, 28, 43, 31, 35, 40, 38, 33, 30, 42
]
```

```python
df = pd.DataFrame({
    'Delivery_Time': delivery_times
})
```

```python
print(df.head())
```

```python
sns.histplot(data=df, x='Delivery_Time', bins=8, kde=True)

plt.title('Distribution of Swiggy Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Number of Orders')

plt.show()
```

TASK:2.
Download or create a list of daily step counts (e.g., from a fitness app) for one month and plot the distribution using sns.kdeplot(), showing the probability density curve

```python
steps = [
    4500, 6200, 7800, 5200, 8900, 10200, 7500, 6800, 9400, 8300,
    5600, 7200, 11000, 9800, 6400, 8700, 9200, 7100, 5900, 10500,
    8100, 7600, 6800, 9700, 11500, 8400, 6300, 9000, 7300, 10000
]
```

```python
df = pd.DataFrame({'Daily_Steps': steps})
```

```python
print(df.head())
```

```python
sns.kdeplot(data=df, x='Daily_Steps', fill=True)

plt.title('Distribution of Daily Step Counts')
plt.xlabel('Daily Steps')
plt.ylabel('Probability Density')

plt.show()
```

TASK:3.
Combine sns.histplot() and sns.kdeplot() to visualize the distribution of monthly spending amounts (in rupees) on Zomato orders, using the same dataset for both plots on a single figure.<br><br><em><strong>Hint:</strong> Use the same data and overlay the KDE curve on the histogram by calling both functions on the same axes.</em>

```python
spending = [
    450, 600, 750, 500, 900, 1200, 850, 650, 1500, 1100,
    700, 950, 1300, 800, 550, 1600, 1000, 750, 1400, 900,
    1800, 1250, 600, 1050, 2000, 850, 1150, 700, 1450, 950
]
```

```python
df = pd.DataFrame({'Monthly_Spending': spending})
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='Monthly_Spending',
    bins=8,
    kde=True,
    stat='density'
)

plt.title('Distribution of Monthly Zomato Spending')
plt.xlabel('Monthly Spending (₹)')
plt.ylabel('Probability Density')

plt.show()
```

TASK:4.
Given a dataset of IPL match scores, use sns.histplot() to plot the run distribution. Change the number of bins and observe how the plot changes. Write 2-3 lines about what you notice.<br><br><em><strong>Hint:</strong> Try bins=5, bins=10, and bins=20.</em>

```python
scores = [
    145, 178, 192, 156, 210, 165, 134, 185, 201, 172,
    148, 220, 195, 160, 182, 175, 140, 205, 188, 167,
    153, 198, 215, 130, 176, 190, 158, 208, 180, 169,
    143, 187, 200, 162, 193, 155, 212, 174, 184, 146,
    197, 170, 225, 159, 181, 204, 138, 191, 166, 178
]
```

```python
df = pd.DataFrame({'Runs': scores})
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(data=df, x='Runs', bins=5, ax=axes[0])
axes[0].set_title('IPL Scores (5 Bins)')
axes[0].set_xlabel('Runs')
axes[0].set_ylabel('Frequency')

sns.histplot(data=df, x='Runs', bins=10, ax=axes[1])
axes[1].set_title('IPL Scores (10 Bins)')
axes[1].set_xlabel('Runs')
axes[1].set_ylabel('Frequency')

sns.histplot(data=df, x='Runs', bins=20, ax=axes[2])
axes[2].set_title('IPL Scores (20 Bins)')
axes[2].set_xlabel('Runs')
axes[2].set_ylabel('Frequency')

plt.tight_layout()
plt.show()

```
