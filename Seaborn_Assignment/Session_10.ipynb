TASK:1.
Use sns.set() to apply the 'poster' context to a Seaborn barplot showing the number of songs streamed per day on Spotify, and observe how the plot elements change compared to the default context.

```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = {
    'Day': ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
            'Friday', 'Saturday', 'Sunday'],
    'Songs_Streamed': [25, 30, 22, 35, 40, 55, 48]
}
```

```python
df = pd.DataFrame(data)
```

```python
sns.set(context='poster')

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Day', y='Songs_Streamed', color='green')

plt.title('Spotify Songs Streamed Per Day')
plt.xlabel('Day')
plt.ylabel('Number of Songs Streamed')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

TASK:2.
Create a Seaborn countplot showing the distribution of food orders by cuisine type (e.g., Indian, Chinese, Italian) using a 'colorblind' palette, similar to how Zomato might visualize order trends.<br><br><em><strong>Hint:</strong> Use palette='colorblind' in your plot function.</em>

```python
data = {
    'Cuisine': [
        'Indian', 'Chinese', 'Italian', 'Indian', 'Mexican',
        'Chinese', 'Indian', 'Italian', 'Indian', 'Chinese',
        'Mexican', 'Italian', 'Indian', 'Chinese', 'Indian',
        'Italian', 'Chinese', 'Indian', 'Mexican', 'Italian'
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x='Cuisine', palette='colorblind', hue='Cuisine', legend=False)

plt.title('Distribution of Food Orders by Cuisine')
plt.xlabel('Cuisine Type')
plt.ylabel('Number of Orders')
plt.tight_layout()
plt.show()
```

TASK:3.
Generate a Seaborn lineplot of daily step counts (simulate data for 2 weeks) and style it using the 'muted' palette and 'notebook' context for a clean, presentation-ready look.

```python
data = {
    'Day': list(range(1, 15)),
    'Steps': [4500, 6200, 5800, 7000, 8500, 9200, 7500,
              6800, 8000, 9500, 8800, 10500, 9800, 11000]
}

df = pd.DataFrame(data)
```

```python
sns.set_theme(context='notebook', palette='muted')

plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x='Day', y='Steps', marker='o')

plt.title('Daily Step Counts Over Two Weeks')
plt.xlabel('Day')
plt.ylabel('Number of Steps')
plt.xticks(range(1, 15))
plt.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
data = {
    'Cuisine': ['Indian', 'Chinese', 'Italian', 'Mexican', 'Indian',
                'Chinese', 'Italian', 'Mexican', 'Indian', 'Chinese'],
    'Orders': [50, 35, 40, 25, 60, 30, 45, 20, 55, 38]
}

df = pd.DataFrame(data)
```

TASK:5.
Design a Seaborn theme that mimics the look of a Flipkart sales dashboard: use sns.set() to set a 'deep' palette, 'talk' context, and customize the font scale and grid style. Apply your theme to a barplot showing sales of 5 product categories.

```python
data = {
    'Product Category': [
        'Electronics', 'Clothing', 'Groceries',
        'Home Appliances', 'Footwear'
    ],
    'Sales': [85000, 60000, 45000, 70000, 35000]
}

df = pd.DataFrame(data)
```

```python
sns.set(
    palette='deep',
    context='talk',
    font_scale=0.8,
    style='whitegrid'
)

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=df,
    x='Product Category',
    y='Sales'
)

plt.title('Flipkart Sales Dashboard', fontweight='bold')
plt.xlabel('Product Category')
plt.ylabel('Sales Amount (₹)')
plt.xticks(rotation=20)

for container in ax.containers:
    ax.bar_label(container, fmt='₹%.0f', padding=5, fontsize=10)

plt.tight_layout()
plt.show()
```
