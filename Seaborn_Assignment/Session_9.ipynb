TASK:1.
Use Seaborn's regplot to visualize the relationship between the number of likes and number of comments on a set of Instagram posts (create a small DataFrame with at least 10 sample data points).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    'Post': [
        'Post 1', 'Post 2', 'Post 3', 'Post 4',
        'Post 5', 'Post 6', 'Post 7', 'Post 8',
        'Post 9', 'Post 10', 'Post 11', 'Post 12'
    ],
    'Likes': [
        120, 250, 180, 400, 350, 500,
        600, 450, 700, 800, 650, 900
    ],
    'Comments': [
        12, 25, 18, 40, 32, 55,
        60, 42, 75, 85, 65, 95
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
plt.figure(figsize=(9, 5))

sns.regplot(
    data=df,
    x='Likes',
    y='Comments',
    color='blue',
    scatter_kws={'s': 70},
    line_kws={'color': 'red'}
)

plt.title('Instagram Likes vs Comments')
plt.xlabel('Number of Likes')
plt.ylabel('Number of Comments')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

TASK:2.
Create a lmplot in Seaborn to show the regression line between food order price and delivery time using a sample Zomato-style dataset (minimum 15 rows with random values for price and delivery time).

```python
import random
```

```python
random.seed(42)
```

```python
data = {
    'Order_Price': [random.randint(100, 1000) for _ in range(20)],
    'Delivery_Time': [random.randint(15, 60) for _ in range(20)]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Zomato Food Orders Dataset:")
print(df)
```

```python
sns.lmplot(
    data=df,
    x='Order_Price',
    y='Delivery_Time',
    height=5,
    aspect=1.4,
    scatter_kws={'s': 70, 'color': 'blue'},
    line_kws={'color': 'red'}
)

plt.title('Food Order Price vs Delivery Time')
plt.xlabel('Order Price (₹)')
plt.ylabel('Delivery Time (minutes)')
plt.show()
```

TASK:5.
Modify a given lmplot that shows regression between Flipkart product price and number of reviews. Add a title, axis labels, and customize the marker color.<br><br><em><strong>Constraint:</strong> Do not change the regression logic—only update plot aesthetics.</em>

```python
data = {
    'Price': [299, 499, 799, 999, 1299, 1599, 1999, 2499, 2999, 3499],
    'Reviews': [120, 200, 180, 350, 400, 320, 500, 450, 600, 550]
}

```

```python
df = pd.DataFrame(data)
```

```python
g = sns.lmplot(
    data=df,
    x='Price',
    y='Reviews',
    scatter_kws={'color': 'purple'}
)

g.set_axis_labels('Product Price (₹)', 'Number of Reviews')
g.fig.suptitle('Flipkart Product Price vs Number of Reviews', y=1.03)

plt.show()
```
